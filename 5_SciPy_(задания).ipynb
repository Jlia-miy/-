```python
# импорт необходимых библиотек
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
```

## Задание 1. Экспоненциальное распределение

Постройте график функции плотности вероятности (PDF) для экспоненциального распределения с параметром $\lambda=0.5.$

```python
x = np.linspace(-5, 30, 1000)
lamda = 0.5
pdf = stats.expon.pdf(x, loc=0, scale=1/lamda)
plt.plot(x, pdf, label="PDF")
plt.title('PDF для экспоненциального распределения')
plt.xlabel('x')
plt.ylabel('P(x)')
plt.legend()
plt.grid(True)
plt.show()
```

Постройте график функции распределения (CDF) для экспоненциального распределения с параметром $\lambda=0.5.$

```python
x = np.linspace(-5, 30, 1000)
lamda = 0.5
cdf = stats.expon.cdf(x, loc=0, scale=1/0.5)

# Построение графика CDF
plt.plot(x, cdf, label="CDF", color='y')
plt.title('CDF для экспоненциального распределения')
plt.xlabel('x')
plt.ylabel('F(x)')
plt.legend()
plt.grid(True)
plt.show()
```

Сгенерируйте 1000 случайных чисел из экспоненциального распределения с параметром $\lambda=0.5,$ затем постройте гистограмму и наложите на неё теоретическую функцию плотности вероятности.

```python
# Генерация случайных чисел для нормального распределения
lamda = 0.5
random_data = stats.expon.rvs(loc=0, scale=1/lamda, size=1000)

# Построение гистограммы случайных чисел
plt.hist(random_data, bins=30, density=True, alpha=0.6, color='r')

# Наложение теоретической плотности
x = np.linspace(-5, 5, 1000)
pdf = stats.expon.pdf(x, loc=0, scale=1)
plt.plot(x, pdf, 'r', linewidth=2)

plt.title('Гистограмма и PDF для экспоненциального распределения')
plt.show()
```

## Задание 2. Биномиальное распределение


Постройте график функции вероятности (PMF) для биномиального распределения с параметрами $n=10$ и $p=0.3.$

```python

x = np.arange(0, 11)
pmf = stats.binom.pmf(x, n=10, p=0.3)

# Построение графика PMF
plt.stem(x, pmf, label="PMF", basefmt=" ")
plt.title('Функция вероятности для биномиального распределения')
plt.xlabel('x')
plt.ylabel('P(x)')
plt.legend()
plt.grid(True)
plt.show()
```

Постройте график функции распределения (CDF) для биномиального распределения с параметрами $n=10$ и $p=0.3.$

```python
x = np.linspace(-5, 20, 1000)
cdf = stats.binom.cdf(x, 10, 0.3)

# Построение графика CDF
plt.plot(x, cdf, label="CDF", color='y')
plt.title('Функция распределения для биномиального распределения')
plt.xlabel('x')
plt.ylabel('F(x)')
plt.legend()
plt.grid(True)
plt.show()
```

Сгенерируйте 1000 случайных чисел из биномиального распределения с параметрами $n=10$ и $p=0.3,$ затем постройте гистограмму и наложите на неё теоретическую функцию вероятности.

```python
# Генерация случайных чисел для нормального распределения
random_data = stats.binom.rvs(10, 0.3, size=1000)

# Построение гистограммы случайных чисел
plt.hist(random_data, bins=30, density=True, alpha=0.6, color='g')

# Наложение теоретической плотности
x = np.linspace(-5, 5, 1000)
pdf = stats.binom.cdf(x, 10, 0.3)
plt.plot(x, pdf, 'r', linewidth=2)

plt.title('Гистограмма и PDF для биномиального распределения')
plt.show()
```

## Задание 4. Нормальное распределение

Дано нормальное распределение с параметрами:
среднее μ=0,
стандартное отклонение σ=1.

Постройте график плотности вероятности (PDF) и рядом график функции распределения (CDF) для нормального распределения (использовать subplot).

```python
x = np.linspace(-5, 5, 1000)

pdf = stats.norm.pdf(x, loc=0, scale=1)

plt.plot(x, pdf, label="PDF")
plt.title('Функция плотности вероятности для нормального распределения')
plt.xlabel('x')
plt.ylabel('P(x)')
plt.legend()
plt.grid(True)
plt.show()
```

Добавьте на график плотности вероятности (PDF) гистограмму выборки (1000 случайных значений из этого распределения).

```python
# Генерация случайных чисел для нормального распределения
random_data = stats.norm.rvs(loc=0, scale=1, size=1000)

# Построение гистограммы случайных чисел
plt.hist(random_data, bins=30, density=True, alpha=0.6, color='y')

# Наложение теоретической плотности
x = np.linspace(-5, 5, 1000)
pdf = stats.norm.pdf(x, loc=0, scale=1)
plt.plot(x, pdf, 'r', linewidth=2)

plt.title('Гистограмма и PDF для нормального распределения')
plt.show()
```

Добавьте вертикальные линии для квантилей 5%, 50% и 95%.

```python
# Генерация случайных чисел для нормального распределения
random_data = stats.norm.rvs(loc=0, scale=1, size=1000)

# Построение гистограммы случайных чисел
plt.hist(random_data, bins=30, density=True, alpha=0.6, color='y')

# Наложение теоретической плотности
x = np.linspace(-5, 5, 1000)
pdf = stats.norm.pdf(x, loc=0, scale=1)
plt.plot(x, pdf, 'r', linewidth=2)

probabilities = [0.05, 0.5, 0.95]
ppf_values = stats.norm.ppf(probabilities, loc=0, scale=1)
plt.vlines(x=ppf_values, ymin=0, ymax=0.7, colors=['blue', 'green', 'purple'],
           linestyles='dashed', linewidth=1.5)

plt.title('Гистограмма и PDF для нормального распределения')
plt.show()
```

## Задание 5. Построение Q-Q Plot

5.1. Сгенерируйте три выборки данных и постройте для каждой Q-Q график для проверки нормальности распределения:

Выборка A: 200 чисел из нормального распределения (mean=5, std=2)

Выборка B: 200 чисел из равномерного распределения (от 0 до 10)

Выборка C: 200 чисел из экспоненциального распределения (scale=1.5)

```python
# Нормальное распределение
np.random.seed(42)
data = stats.norm.rvs(loc=5, scale=2, size=200)

# Создание Q-Q графика
fig, ax = plt.subplots(figsize=(8, 6))
res = stats.probplot(data, dist="norm", plot=ax)
ax.set_title('Q-Q Plot')
plt.show()
```

```python
# Равномерное распределение
np.random.seed(42)
data = stats.uniform.rvs(loc=0, scale=10, size=200)

# Создание Q-Q графика
fig, ax = plt.subplots(figsize=(8, 6))
res = stats.probplot(data, dist="uniform", plot=ax)
ax.set_title('Q-Q Plot')
plt.show()
```

```python
# Экспоненциальное распределение
np.random.seed(42)
data = stats.expon.rvs(scale=1.5, size=200)

# Создание Q-Q графика
fig, ax = plt.subplots(figsize=(8, 6))
res = stats.probplot(data, dist="expon", plot=ax)
ax.set_title('Q-Q Plot')
plt.show()
```
