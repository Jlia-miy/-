# Задание 5. Работа с пропущенными значениями

Создайте DataFrame с пропущенными значениями:

```python
data_with_na = {
    'Имя': ['Петр', 'Евгения', 'Алексей', None],
    'Возраст': [25, None, 22, 29],
    'Город': ['Москва', 'Казань', 'Москва', 'Новосибирск']
}
new_data = pd.DataFrame(data_with_na)
new_data
# продолжите здесь
```

Найдите количество пропущенных значений в каждом столбце.

```python
new_data.isnull().sum()
```

Заполните пропущенные значения в столбце "Возраст" средним значением.

```python
new_data['Возраст'].fillna(new_data['Возраст'].mean(), inplace=True)
new_data
```

# Задание 6. Слияние и объединение DataFrame

Используя два DataFrame:

```python
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Имя': ['Степан', 'Инна', 'Алексей']
})

df2 = pd.DataFrame({
    'ID': [2, 3, 4],
    'Город': ['Москва', 'Казань', 'Новосибирск']
})
```

```python
df1
```

```python
df2
```

Объедините два DataFrame по столбцу "ID".

```python
df1.merge(df2[['ID', 'Город']], on='ID')
```

Выполните слияние по внешнему ключу (outer join).

```python
df1.merge(df2[['ID', 'Город']], how='outer')
```

```python
max(event['Дата'])
```

# Задание 8. Сортировка данных
