## 1. Гистограммы и распределения

Загрузите встроенный набор данных penguins и постройте гистограмму распределения массы тела (body_mass_g).

```python
df = pd.DataFrame(sns.load_dataset("penguins"))

plt.hist(df['body_mass_g'], bins=36, color='LightCyan', edgecolor='DarkOliveGreen')
plt.title("Гистограмма распределения массы тела")
plt.xlabel("Масса тела, g")
plt.ylabel("Количество людей")
plt.show()
```
