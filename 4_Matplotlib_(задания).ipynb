```python
# Импорт необходимых библиотек
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import pandas as pd
import numpy as np
```

##Задание 1. Основы Matplotlib


Постройте график синусоиды от 0 до $2\pi$ с помощью matplotlib.

Настройте оси (подписи, сетка) и добавьте легенду.

Измените цвет линии на красный и стиль линии на пунктирный.

```python

x = np.linspace(0, 2*np.pi, 500)
y = np.sin(x)


fig, ax = plt.subplots()
ax.plot(x, y, color='red', linestyle='--', linewidth=1, label=r"$\sin(x)$")
ax.set(xlim=(-1, 2*np.pi+1), xlabel="x")
ax.set(ylim=(-1, 1), ylabel="y")
ax.legend(fontsize=10)
plt.show()
```

##Задание 2. Гистограмма

```python
Создайте случайные данные (например, 1000 случайных значений из нормального распределения).

Постройте гистограмму и настройте количество бинов.

Измените цвет столбцов на синий и добавьте сетку.
```

```python
mu, sigma = 0, 1 #среднее и стандартное отклонение
data = np.random.normal(mu, sigma, 1000)


plt.hist(data, bins=36, color='skyblue', edgecolor='yellow')

plt.title("Гистограмма нормального распределения")
plt.xlabel("Значения")
plt.ylabel("Плотность вероятности")
plt.show()
```

##Задание 3. Подграфики

Создайте два графика: один с функцией синуса, второй — с функцией косинуса.

Разместите их в одном окне (с использованием plt.subplot()).

```python

x = np.linspace(0, 2*np.pi, 500)
y1 = np.sin(x)
y2 = np.cos(x)

fig, ax = plt.subplots()
'''
Фигуру (fig) — контейнер верхнего уровня, "холст", на котором располагаются графики
Оси (ax) — собственно область, где рисуется график
'''
ax.plot(x, y1, color='red', linestyle='--', linewidth=1, label=r"$\sin(x)$")
ax.plot(x, y2, color='blue', linestyle=':', linewidth=1, label=r"$\cos(x)$")
ax.set(xlim=(-1, 2*np.pi+1), xlabel="x")
ax.set(ylim=(-1, 1), ylabel="y")
ax.legend(fontsize=10)
plt.show()


```
