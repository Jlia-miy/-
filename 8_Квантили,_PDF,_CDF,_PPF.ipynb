# Задание
Для каждого из заданных ниже распределений:

1. Найти квантили 5%, 25%, 50%, 90%, 95%
2. Построить в одном окне (используя subplot) графики:
- функции плотности (PDF),
- функции распределения (CDF),
- обратной функции распределения (PPF)
3. Отметить квантили на графиках

```python
#Бибилиотеки
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import math
```

```python
X = np.linspace(-5, 5, 1000) #значение по x
```

```python
#ВЫЧИСЛЕНИЕ КВАНТИЛЕЙ
q_kfantily = [0.05, 0.25, 0.50, 0.90, 0.95]
Q = []
n = len(X)
for q in q_kfantily:
  koef = (n-1)*q
  if koef == math.floor(koef):
    Q.append(X[(n-1)*q])
  else:
    Q.append((X[math.floor((n-1)*q)] + X[math.floor((n-1)*q)+1])/2)
#----------------------------------------------------------------------
```

## Стандартное нормальное распределение $\mathcal{N}(0,1)$

```python

pdf = stats.norm.pdf(X, loc=0, scale=1)
cdf = stats.norm.cdf(X, loc=0, scale=1)
ppf = stats.norm.ppf(X, loc=0, scale=1)
#----------------------------------------------------------------------
for value_x in Q:
  plt.axvline(x=round(value_x, 3), color='MediumBlue', linestyle=':', linewidth=1.5, alpha=0.3)

plt.xticks(ticks=Q,
           labels=[f'Q({kfantily})' for kfantily in q_kfantily])
#----------------------------------------------------------------------

plt.plot(X, pdf, label="PDF")
plt.plot(X, cdf, label="CDF", color='r')
plt.plot(X, ppf, label="PPF", color='g')
plt.title('НОРМАЛЬНОЕ РАСПРЕДЕЛЕНИЕ')
plt.xlabel('x')
plt.ylabel('P(x)')
plt.legend()
plt.grid(True)
plt.show()
```

## Экспоненциальное распределение с $\lambda=3$

```python
lamda = 0.5
pdf = stats.expon.pdf(X, loc=0, scale=1/lamda)
cdf = stats.expon.cdf(X, loc=0, scale=1/lamda)
ppf = stats.expon.ppf(X, loc=0, scale=1/lamda)
#----------------------------------------------------------------------
for value_x in Q:
  plt.axvline(x=round(value_x, 3), color='MediumBlue', linestyle=':', linewidth=1.5, alpha=0.3)

plt.xticks(ticks=Q,
           labels=[f'Q({kfantily})' for kfantily in q_kfantily])
#----------------------------------------------------------------------

plt.plot(X, pdf, label="PDF")
plt.plot(X, cdf, label="CDF", color='r')
plt.plot(X, ppf, label="PPF", color='g')
plt.title('ЭКСПОНЕЦИАЛЬНОЕ РАСПРЕДЕЛЕНИЕ')
plt.xlabel('x')
plt.ylabel('P(x)')
plt.legend()
plt.grid(True)
plt.show()
```
